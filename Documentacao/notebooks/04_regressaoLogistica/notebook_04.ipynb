## Criando a primeira base de dados e visualizando suas amostras

Primeiro, iremos importar todas as bibliotecas que serão usadas ao longo deste notebook.

```python
# -*- coding: utf-8 -*-
import numpy as np # importa a biblioteca usada para trabalhar com vetores e matrizes
import pandas as pd # importa a biblioteca usada para trabalhar com dataframes

import sklearn as skl
from sklearn import datasets
from sklearn import preprocessing
from sklearn import linear_model
```

Iremos criar uma base de dados onde as amostras são separáveis linearmente. Para isso, iremos usar a classe *make_classification* da biblioteca *sklearn*. 

```python
# gerar dados de classificação
X, Y = skl.datasets.make_classification(n_samples=500, n_features=2, n_classes=2, n_informative=2, 
                                        n_redundant=0, n_clusters_per_class=1, class_sep = 2.3,  
                                        flip_y = 0.0, random_state=10)

# imprime as 5 primeiras linhas da matriz X
display('X:', X[0:5,:])

# imprime os 5 primeiros valores de Y
print('Y:', Y[0:5])
```

Agora, vamos criar uma função para plotar os dados.

```python
# importa a biblioteca matplotlib
import matplotlib.pyplot as plt

# função para plotar os dados
def visualizarDados(X,Y):
    """
    Função usada para plotar os dados
    """
    
    # define o tamanho da figura 
    plt.figure(figsize=(10,8))
    
    # plota os dados da classe 0
    plt.scatter( X[Y==0,0], X[Y==0,1], label='Classe 1', color='blue', s=50, edgecolors='k') 
    
    # plota os dados da classe 1
    plt.scatter( X[Y==1,0], X[Y==1,1], label='Classe 0', color='red', s=50, edgecolors='k') 
        
    # plota a legenda
    plt.legend()
    
# chama a função que plota os dados   
visualizarDados(X,Y)
plt.show()
```

---
## Treinando a regressão logística

Vamos treinar um modelo de classificação usando o método de regressão logística. Para isso, iremos empregar o método [LogisticRegression](https://scikit-learn.org/1.5/modules/generated/sklearn.linear_model.LogisticRegression.html) da biblioteca **scikit-learn**.

Iremos definir o parâmetro de regularização com um valor de custo igual a 1.0. Na biblioteca *scikit learn*, o custo da regressão logística é um parâmetro que deve ter valor positivo e que é inverso à força de regularização. Quanto menor o valor desse parâmetro, maior é a regularização.

```python
# inicia o classificador 
model = skl.linear_model.LogisticRegression(C=1.0, random_state=10)

# treina o classificador com os dados de treinameto
model.fit(X, Y) 

print('Informações do modelo: ')
print(model)

print('\nPesos encontrados: ')
print(model.coef_) 
```

Agora, vamos exibir o limite de decisão gerado pela regressão logística.

```python
def plota_limite_decisao(model, X, Y, ax):
    """
    Plota o limite de decisão
    """
   
    x = X[:, 0]
    y = X[:, 1]
    
    x_min, x_max = x.min() - 1, x.max() + 1
    y_min, y_max = y.min() - 1, y.max() + 1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.01),np.arange(y_min, y_max, 0.01))
    
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = np.array(Z)
    Z = Z.reshape(xx.shape)
    out = ax.contourf(xx, yy, Z, cmap=plt.cm.coolwarm, alpha=0.8)

    ax.set_xlim(x.min()-0.1, x.max()+0.1)
    ax.set_ylim(y.min()-0.1, y.max()+0.1)
    ax.set_xticks(())
    ax.set_yticks(())

    #plota os dados da classe 0
    plt.scatter( X[Y==0,0], X[Y==0,1], label='Classe 1', color='blue', s=50, edgecolors='k') 
    
    #plota os dados da classe 1
    plt.scatter( X[Y==1,0], X[Y==1,1], label='Classe 0', color='red', s=50, edgecolors='k') 

    #insere a legenda
    plt.legend()


# define o tamanho da figura
fig, ax = plt.subplots(figsize=(8, 8)) 

# plota o limite de decisão
plota_limite_decisao(model, X, Y, ax)
plt.show()
```

O gráfico acima apresenta o limite de decisão criado pelo modelo com custo igual a 1. Veja que com esse valor do parâmetro C, o modelo errou a predição de dois dados. 


Agora, vamos gerar outro modelo, mas usando custo igual a 100.

```python
# inicia o classificador 
model = skl.linear_model.LogisticRegression(C=100.0, random_state=10)

# treina o classificador com os dados de treinameto
model.fit(X, Y) 

# define o tamanho da figura
fig, ax = plt.subplots(figsize=(8, 8)) 

# plota o limite de decisão
plota_limite_decisao(model, X, Y, ax)
plt.show()
```

Conforme pode ser observado nos exemplos acima, quando usamos valores altos para o parâmetro de custo, a regressão logística tenta classificar todos os exemplos de treinamento corretamente (baixa regularização)

Antes de ir para a próxima etapa, você pode testar outros valores de custo e verificar como isso afeta o limite de decisão.

## Regressão logística aplicada em dados não separáveis linearmente

Antes de implementar a regressão logística, iremos criar uma base de dados com amostras que não são linearmente separáveis.

```python
# Gera uma base de dados com dois círculos
X2, Y2 = skl.datasets.make_circles(n_samples=1000, noise=0.2, 
                                   factor=0.1, random_state = 10)

# imprime as 5 primeiras linhas da matriz X
display('X:', X2[0:5,:])

# imprime os 5 primeiros valores de Y
print('Y:', Y2[0:5])

# plota os dados de treinamento
visualizarDados(X2,Y2)
plt.show()
```

Claramente, as amostras não são linearmente separáveis e, portanto, novos atributos polinomiais precisarão ser criados para melhorar o limite de decisão da regressão logística.

Abaixo, vamos construir uma função que gera dados polinomiais. Esta função mapeia o vetor original com apenas duas colunas em um vetor transformado com 27 dimensões. Dessa forma, o classificador gerado pela regressão logística será treinado com uma base de dimensão mais elevada e poderá produzir um limite de decisão mais complexo.

```python
poly = skl.preprocessing.PolynomialFeatures(degree=6, include_bias=False)

# atributosPolinomiais adiciona novas colunas que correspondem a atributos polinomiais
X2_poli = poly.fit_transform(X2)

print('Dimensão do novo conjunto de dados polinomiais: \n', X2_poli.shape)
```

Se por um lado o mapeamento dos atributos pode aumentar a robustez do classificador, por outro, ele ficará mais suscetível ao super-ajustamento dos dados (*overfitting*). Assim sendo, para resolver esse impasse, iremos implementar a regressão logística com regularização.

Vamos então treinar dois modelos de regressão logística com as amostras dessa segunda base de dados. O primeiro irá usar um valor baixo de custo (alta regularização), enquanto o segundo irá usar um valor alto de custo. 

```python
# inicia o classificador 
model1 = skl.linear_model.LogisticRegression(C=0.01, random_state=10)
model2 = skl.linear_model.LogisticRegression(C=100, random_state=10)

# treina o classificador com os dados de treinameto
model1.fit(X2_poli, Y2) 
model2.fit(X2_poli, Y2) 

print("Modelos treinados com sucesso!")
```

Abaixo é mostrado o limite de decisão de cada classificador. 

```python
def plotarLDNaoLinear(model, title = ""):
    
    # plota os dados de treinamento
    visualizarDados(X2,Y2)

    u = np.linspace(-1, 1.5, 50)
    v = np.linspace(-1, 1.5, 50)

    z = np.zeros( [len(u), len(v)] )

    for i in range( len(u) ):
        for j in range( len(v) ):
            
            newData = np.array( [[u[i], v[j]]] )
            poliTransf = poly.transform( newData )
            z[i,j] = model.predict( poliTransf );

    plt.contour(u, v, z, levels=[0.5], cmap="viridis")
    
    plt.title(title)

    plt.show()
    

# plota o modelo
plotarLDNaoLinear(model1, title = "Modelo com C=0.01")  

# plota o modelo
plotarLDNaoLinear(model2, title = "Modelo com C=100")  
```

## Conclusão

Neste notebook, exploramos o comportamento da regressão logística em diferentes cenários de dados, observando como o parâmetro de regularização influencia o desempenho do modelo. Inicialmente, aplicamos a regressão logística a uma base de dados linearmente separável, onde foi possível observar que valores maiores de $C$ (inverso da regularização) permitiram ao modelo ajustar-se de forma mais rígida aos dados, resultando em classificações mais precisas. No entanto, também foi constatado que um $C$ muito elevado pode levar ao *overfitting*, comprometendo a generalização do modelo.

Em seguida, ao trabalhar com dados não separáveis linearmente, utilizamos atributos polinomiais para transformar o espaço de entrada. Nesse cenário mais complexo, o ajuste cuidadoso do parâmetro $C$ tornou-se ainda mais importante, uma vez que um valor inadequado pode resultar *underfitting* ou em *overfitting*.

Esses experimentos ilustram a flexibilidade da regressão logística para lidar com diferentes tipos de dados e destacam a necessidade de ajustar adequadamente seus parâmetros de acordo com os dados e a complexidade do problema.
