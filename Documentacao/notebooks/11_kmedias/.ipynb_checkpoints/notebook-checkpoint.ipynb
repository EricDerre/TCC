## Introdução

Neste notebook, iremos explorar o método de K-médias (Lloyd, 1982), um algoritmo de agrupamento amplamente utilizado para identificar padrões e separar dados em grupos distintos. Iremos criar uma base de dados sintética com conjuntos circulares e veremos como o K-médias se ajusta os centros dos grupos e define o limite de decisão entre eles. Por fim, demonstraremos o uso do K-médias para compressão de imagens, agrupando pixels semelhantes e reduzindo o número de cores da imagem. Esse exemplo mostra como o algoritmo pode ser aplicado em problemas práticos de simplificação e visualização de dados.

Este *notebook* se refere aos Capítulos **12 (Análise de Agrupamentos)** e **13 (Algoritmos de Agrupamento)**.

## Criando a base de dados e visualizando suas amostras

Primeiro, iremos importar todas as bibliotecas que serão usadas ao longo deste notebook.

```python
# -*- coding: utf-8 -*-
import numpy as np #importa a biblioteca usada para trabalhar com vetores e matrizes
import pandas as pd #importa a biblioteca usada para trabalhar com dataframes
import cv2 # função para ler dados em formato de imagem

import sklearn as skl
from sklearn import cluster
from sklearn import datasets
```

Iremos criar uma base de dados onde as amostras são separáveis linearmente. Para isso, iremos usar a classe *make_classification* da biblioteca *sklearn*. 

```python
# Gerar dados 
X, _ = skl.datasets.make_blobs(n_samples=500, centers=3, n_features=2, 
                  cluster_std=0.8, center_box=(-6.5, 6.5), random_state=10)

# Imprime as 5 primeiras linhas da matriz X
display('X:', X[0:5,:])
```

Agora, vamos criar uma função para plotar os dados.

```python
# importando a biblioteca matplotlib
import matplotlib.pyplot as plt

# função para plotar os dados
def visualizarDados(X,Y):
    """
    Função usada para plotar os dados
    """
    
    # definindo o tamanho da figura 
    plt.figure(figsize=(10,6))
    
    # plota os dados
    plt.scatter( X, Y, marker='o', label = 'Dados', color='blue', s=50, edgecolors='k') 
    
    # define os nomes do eixo x e do eixo y
    plt.xlabel(r'F1',fontsize='medium') 
    plt.ylabel(r'F2',fontsize='medium') 

    #define o título do gráfico
    plt.title('Base de dados', fontsize='x-large')
    
# chamando a função que plota os dados   
visualizarDados(X[:,0],X[:,1])
plt.show()
```

Agora, iremos aplicar o algoritmo K-médias para realizar o agrupamento dos dados. Utilizamos um modelo com 3 grupos, até um máximo de 500 iterações e uma semente aleatória para garantir a reprodutibilidade. Após ajustar o modelo com os dados, obtemos os rótulos dos grupos para cada ponto e fazemos previsões para determinar a qual grupo cada ponto pertence. Por fim, extraímos as coordenadas dos centróides de cada cluster.

```python
def executa_kmedias(X, n_clusters = 2, seed = 10):
    """
    Faz todo o processo de treinamento e predição
    """
    
    # Inicializa o modelo
    modelo =  skl.cluster.KMeans(n_clusters=n_clusters, max_iter=500, 
                                     n_init='auto', random_state = seed)

    # Treina o modelo nos dados
    modelo.fit(X)

    # Obtém os rótulos de cada dado (grupo ao qual ele pertence)
    pred  = modelo.predict(X)
    
    return modelo, pred
  
# executa o K-médias com 3 grupos
modelo, pred = executa_kmedias(X, n_clusters = 3, seed = 10)

print("Modelo: \n", modelo)

print("\nCentróides: \n", modelo.cluster_centers_)

print("\nCusto: \n", modelo.inertia_)

print("\nGrupos dos 5 primeiros dados:", pred[0:5])
```

Iremos plotar o limite de decisão aprendido pelo modelo.

```python
#função para plotar o limite de decisão
def plota_limite_decisao(model, X, ax):
    """
    Função usada para plotar o modelo de decisão do SVM
    """
   
    x = X[:, 0]
    y = X[:, 1]
    
    x_min, x_max = x.min() - 1, x.max() + 1
    y_min, y_max = y.min() - 1, y.max() + 1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.01),np.arange(y_min, y_max, 0.01))
    
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = np.array(Z)
    Z = Z.reshape(xx.shape)
    out = ax.contourf(xx, yy, Z, cmap=plt.cm.coolwarm, alpha=0.8)

    ax.set_xlim(x.min()-0.1, x.max()+0.1)
    ax.set_ylim(y.min()-0.1, y.max()+0.1)
    ax.set_xticks(())
    ax.set_yticks(())

    #plota os dados da classe 0
    plt.scatter( X[:,0], X[:,1], color='blue', s=50, edgecolors='k') 
    #insere a legenda
    #plt.legend()
    plt.title( "K-médias com K=" + str(len(model.cluster_centers_)) )


# define o tamanho da figura
fig, ax = plt.subplots(figsize=(8, 8)) 

# plota o limite de decisão
plota_limite_decisao(modelo, X, ax)
plt.show()
```

Agora, vamos testar o K-médias para diferentes valores de $k$. 

```python
valores_k = np.array( [1,2,3,4,5,6] )

# inicializa os custos
custo = np.zeros( len(valores_k) )

for i, k in enumerate(valores_k):

        print('\n%s\nK = %d\n%s\n' %(20*'=',i+1,20*'=') )

        # Executa o K-médias com 3 grupos
        modelo, pred = executa_kmedias(X, n_clusters = k, seed = i*10)

        # Obtém o custo
        custo[i] = modelo.inertia_

        # Define o tamanho da figura
        fig, ax = plt.subplots(figsize=(8, 8)) 

        plota_limite_decisao(modelo, X, ax)
        
        plt.show()
```

É possível ver pelos limites de decisão gerados acima que uma má escolha da quantidade de grupos, pode levara agrupamentos de baixa qualidade. Isso é refletido também no custo, que é uma medida que reflete o erro do agrupamento. Abaixo, iremos plotar os valores de custo que foram coletados para cada número de grupos testados anteriormente. 

```python
# Definindo o tamanho da figura 
plt.figure(figsize=(8,8))

plt.plot(valores_k, custo, marker='o')

for i in range( len(custo) ):
    plt.annotate('%1.2f' %custo[i],xy=(valores_k[i]+0.05,custo[i]+0.05), fontsize='large')

xmin, xmax = plt.xlim()   # returna o xlim atual
plt.xlim([xmin,xmax+0.2]) # seta o xlim 

plt.xlabel('K (qtd. de centroides)', fontsize='large') 
plt.ylabel('Custo', fontsize='large')  

plt.show()
```

Pelo gráfico acima, é possível observar que o custo apresenta um comportamento típico de "cotovelo". Inicialmente, o custo diminui rapidamente à medida que o número de clusters aumenta, indicando que o modelo está conseguindo agrupar os dados de forma mais eficiente. No entanto, após certo ponto, essa redução no custo se torna menos significativa, sugerindo que adicionar mais clusters não melhora muito a qualidade do agrupamento. Esse ponto de inflexão, onde a queda desacelera, é conhecido como o "cotovelo" e indica um bom compromisso entre simplicidade e precisão do modelo.

## Usando o K-médias para agrupar os pixels de uma imagem

Iremos usar o $k$-médias para compactar uma imagem, isto é, representá-la usando um número menor de cores. 

Cada pixel da imagem será uma amostra da nossa base de dados. Cada pixel será representado por três valores que variam entre 0 e 255. O primeiro valor representa a intensidade da cor vermelha; o segundo é a intensidade da cor verde; e o terceiro é a intensidade da cor azul (RGB). 

Usando o método $k$-médias, iremos encontrar os agrupamentos das cores dos pixels. Depois, o valor de cada pixel será substituído pelo valor do centróide mais próximo. 

Primeiro, vamos importar a imagem e exibí-la. 

```python
# Importa a imagem
img = cv2.imread('datasets/dog.jpg')

# converte para RGB
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB) 

# Vamos plotar a imagem original
plt.imshow(img); 
plt.title('Imagem original')
plt.show()
```

Iremos transformar a imagem em uma base de dados, onde cada amostra é um pixel da imagem. Além disso, iremos normalizar os valores dos pixels para ficarem entre 0 e 1. 

```python
print('Dimensão inicial: ',  img.shape) # dimensões da matriz de pixels

# Muda a dimensão da matriz de imagens para Nx3 onde N = número de pixels.
# Cada linha contém os valores de vermelho, verde e azul do pixel.
X = np.reshape(np.ravel(img), (img.shape[0]*img.shape[1], 3)) 

print('\nNova dimensão: ',  X.shape)

# Iremos normalizar os valores
X = X / 255 # divide por 255 para que todos os valores fiquem entre 0 e 1

print('\nValores dos 5 primeiros pixels:')
print(X[0:5,:])
```

Agora, vamos plotar os pixels da imagem.

```python
# rotina que plota os pixels
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(10,8))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(X[:,0], X[:,1], X[:,2], marker='.')

plt.show()
```

### Comprimindo a imagem

Agora, usaremos os agrupamentos do K-médias para comprimir a imagem. Para isso, nós iremos primeiramente encontrar os centróides mais próximos de cada pixel. Depois, vamos substituir cada pixel pelo centróide mais próximo.  

```python
# Executa o K-médias com 3 grupos
n_clusters = 4
modelo, pred = executa_kmedias(X, n_clusters = n_clusters, seed = 10)

# Obtém dos centróides
centroids = modelo.cluster_centers_

# Agora vamos mapear amostra da base de dados (pixel) por seu centróide mais próximo.
X_compactado = centroids[pred,:]

print("Custo do agrupamento: ", modelo.inertia_)
```

Iremos voltar a base de dados para a dimensão original da imagem.

```python
# Vamos restaurar as dimensões originais da matriz de pixels
img_compactada = np.reshape( np.ravel(X_compactado), (img.shape[0], img.shape[1], 3 ) )

print("Dimensão da imagem resultante: ", img_compactada.shape)
```

Mostraremos a imagem resultante que possui menos cores. 

```python
# Vamos plotar a imagem original
import matplotlib.pyplot as plt
plt.imshow(img); 
plt.title('Imagem original')
plt.show()

# Vamos plotar a imagem compactada
plt.imshow(img_compactada)
plt.title('Imagem compactada com %d cores.' %n_clusters);
plt.show()
```

## Conclusão

Neste notebook, analisamos o comportamento do K-médias em diferentes configurações de agrupamento, explorando sua capacidade de ajustar os centros dos clusters e definir limites de decisão para separar dados. Ao aplicar o algoritmo em uma base de dados sintética com conjuntos circulares, observamos como o número de clusters (**K**) impacta diretamente a qualidade do agrupamento, com valores inadequados de K levando a agrupamentos menos representativos. A análise da **curva de custo** permitiu identificar o ponto em que o número de clusters resulta em um bom equilíbrio entre simplicidade e precisão.

Além disso, ao utilizar o K-médias para a compressão de imagens, vimos sua eficácia na redução de cores ao agrupar pixels semelhantes, o que demonstra uma aplicação prática em problemas de simplificação de dados visuais. Esses experimentos mostram que o K-médias é uma ferramenta flexível e útil tanto para a exploração de padrões em conjuntos de dados quanto para aplicações de visualização e simplificação.
